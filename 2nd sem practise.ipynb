### To perform additation and removable operation of element in sets.

```python
# list ---------
# add = [append, insert, extend]
# remove = [remove,pop(),clear]
# Set --------- 
# add ={add,update}
# remove = {.remove, discard,pop,clear()}
```

```python
my_set = {1,2,3,4,5,6}
print("\nAdditation using add") 
my_set.add(7)
print(my_set)
print("\n Addition using update")
my_set.update([8,9,10])
print(my_set)
print("\nSubtracting using remove")
my_set.remove(8)
print(my_set)
print("\nSubtracting using discard")
my_set.discard(9)
print(my_set)
print("\nSubtracting using pop")
my_set.pop()
print(my_set)
print("\n Clear the all data")
my_set.clear()
print(my_set)
```

### factorial of the number using function

```python
def factorial(n):
    if n < 0:
        print("factorial is not possible in negative number")
    elif n==1 or n==0 :
        return 1
    else:
        fact = 1
        for i in range (2, n+1):
            fact *=i
        return fact
num = int(input("enter any number"))
result = factorial(num)
print(f"factorial of {num} is ", result)
                
```

### To find GCD of number using recursion function.

```python
def gcd(a,b):
    if b == 0:
        return a
    else:
        return gcd(b,a%b)
num1 = int(input("Enter first number"))
num2 = int(input("Enter second number"))
result = gcd(num1,num2)
print(f"The gcd of {num1} and{num2} is {result}")
```

### OPP to find area and perimeter of rectangle

```python
class Rectangle:
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        return self.length * self.breadth
    def perimeter(self):
        return 2*(self.length + self.breadth)
length = float(input("Enter the length of rectangle"))
breadth = float(input("Enter the breadth of rectangle"))
rect = Rectangle(length,breadth)
print("Area of rectangle is",rect.area())
print("Perimeter of rectangle is",rect.perimeter)
```

### Banking Transation using concept of class and object

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"The deposited amount is {amount} and the new balance = {self.balance}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn amount is {amount} and remaining balance = {self.balance}")
        else:
            print("Insufficient balance")

    def display(self):
        print(f"Account holder name is {self.name}")
        print(f"Current Balance is {self.balance}")


name = input("Enter Account holder name: ")
account = BankAccount(name, 0)   # Initial balance = 0
account.deposit(5000)
account.withdraw(2300)
account.display()
```
